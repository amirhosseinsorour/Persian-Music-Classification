```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import style
```

```python
N_FFTs = [1024, 2048, 4096, 8192, 16384, 32768, 65536]
N_MSs = [t/22.05 for t in N_FFTs]

total_loss = [0.4167, 0.4360, 0.4784, 0.4736, 0.5356, 0.4096, 0.7293]
instrument_loss = [0.0449, 0.0185, 0.1032, 0.032, 0.1153, 0.0878, 0.0963]
scale_loss = [0.3105, 0.3643, 0.3165, 0.3814, 0.3525, 0.2751, 0.5745]

instrument_acc = [99.16, 99.58, 98.66, 99.11, 98.66, 98.96, 98.66]
scale_acc = [93.61, 92.75, 94.37, 93.62, 93.77, 94.07, 91.40]
```

```python
# Plotting Accuracy

plt.plot(N_MSs, instrument_acc, label='Instrument')
plt.plot(N_MSs, scale_acc, label='Scale')

plt.xlabel("Frame Duration (ms)")
plt.ylabel("Accuracy (%)")

plt.xticks(np.arange(0, 31e2, 1e2), minor=True)
plt.yticks(np.arange(90,101,1))
plt.yticks(np.arange(90,101,0.5), minor=True)

plt.legend(loc='center right')
plt.grid(True, which='both')

plt.show()
```

```python
# Plotting Loss

plt.plot(N_MSs, instrument_loss, label='Instrument')
plt.plot(N_MSs, scale_loss, label='Scale')
plt.plot(N_MSs, total_loss, label='Total')

plt.xlabel("Frame Duration (ms)")
plt.ylabel("Loss")

plt.xticks(np.arange(0, 31e2, 1e2), minor=True)
plt.yticks(np.arange(0,1,0.1))
plt.yticks(np.arange(0,1,0.05), minor=True)

plt.legend(loc='upper center')
plt.grid(True, which='both')

plt.show()
```
