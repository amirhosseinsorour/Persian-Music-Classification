```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import style
```

```python
ALPHA = [0.1, 0.3, 0.5, 0.8, 1., 2., 3.]

total_loss = [0.0534, 0.1349, 0.2224, 0.2200, 0.2111, 0.3882, 1.0293]
instrument_loss = [0.0176, 0.0149, 0.0203, 0.0066, 0.0104, 0.0058, 0.0082]
scale_loss = [0.2860, 0.2680, 0.2557, 0.1681, 0.0946, 0.1060, 0.2437]

instrument_acc = [99.85, 99.55, 99.70, 99.85, 99.70, 99.85, 99.85]
scale_acc = [95.55, 93.62, 95.55, 97.18, 98.22, 98.07, 94.96]
```

```python
plt.plot(ALPHA, instrument_acc, label='Instrument')
plt.plot(ALPHA, scale_acc, label='Scale')

plt.xlabel("ALPHA")
plt.ylabel("Accuracy (%)")

plt.xticks(ALPHA)
plt.xticks(np.arange(0, 3.2, 0.1), minor=True)
plt.yticks(np.arange(90,101,1))
plt.yticks(np.arange(90,101,0.5), minor=True)

plt.legend(loc='lower right')
plt.grid(True, which='both')

plt.show()
```

```python
plt.plot(ALPHA, instrument_loss, label='Instrument')
plt.plot(ALPHA, scale_loss, label='Scale')
plt.plot(ALPHA, total_loss, label='Total')

plt.xlabel("ALPHA")
plt.ylabel("Loss")

plt.xticks(ALPHA)
plt.xticks(np.arange(0, 3.2, 0.1), minor=True)
plt.yticks(np.arange(0,1.2,0.1))
plt.yticks(np.arange(0,1.2,0.05), minor=True)

plt.legend(loc='upper center')
plt.grid(True, which='both')

plt.show()
```
